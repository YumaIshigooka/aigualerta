```python
def confidence_interval(data, confidence=0.95):
    n = len(data)
    mu, std = np.mean(data), st.sem(data)
    h = std * st.t.ppf((1 + confidence) / 2., n-1)
    return mu, [mu - h, mu + h]
```

```python
mean, CI = confidence_interval(waterFlow)
```
