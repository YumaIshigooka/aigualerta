```python
#constants
POLICY = 'Pòlissa/Póliza/Policy'
TECHNOLOGY = 'Tecnologia/Tecnología/Technology'
DIAMETER = 'Diàmetre comptador (cm)/Diámetro contador (cm)/Counter diameter (cm)'
USAGE = 'Ús/Uso/Use'
HOUSING = "Tipus d'habitatge/Tipo de vivienda/Type of housing"
DATE = 'Data/Fecha/Date'
CONSUMPTION = 'Índex de lectura (L/h)/Índice de lectura (L/h)/Reading index (L/h)'
```

#### Importar datos

```python
import pandas as pd
import pyarrow.dataset as ds
from sklearn.model_selection import GroupShuffleSplit

file_path = '../data/lectures_horaries_ABD.parquet'

dataset = ds.dataset(file_path, format="parquet")
table = dataset.to_table()
df = table.to_pandas()
```

#### Cambiar proporciones

```python
test_size = 0.2
val_size = 0.2 # Based on the remaining (1 - test_size) data
random_state = 42

print(f"Train size: \t\t{((1 - test_size) * (1 - val_size) * 100):.2f}%")
print(f"Validation size: \t{((1 - test_size) * val_size * 100):.2f}%")
print(f"Train size: \t\t{test_size * 100:.2f}%")
```

#### Dividir los datos
Resultado:
* `train_data`
* `validation_data`
* `test_data`

```python
# Initialize GroupShuffleSplit for testing and temporary sets
gss = GroupShuffleSplit(n_splits=1, train_size=test_size, random_state=random_state)
test_idx, temp_idx = next(gss.split(df, groups=df[POLICY]))

test_data = df.iloc[test_idx]
temp_data = df.iloc[temp_idx]

# Split temp_data for training and vaidation
gss_val_test = GroupShuffleSplit(n_splits=1, train_size=val_size, random_state=random_state)
val_idx, train_idx = next(gss_val_test.split(temp_data, groups=temp_data[POLICY]))

validation_data = temp_data.iloc[val_idx]
train_data = temp_data.iloc[train_idx]

# Resulting datasets
print("Training Data:")
print(train_data.head())
print("\nValidation Data:")
print(validation_data.head())
print("\nTest Data:")
print(test_data.head())
```
