```python
#constants
POLICY = 'Pòlissa/Póliza/Policy'
TECHNOLOGY = 'Tecnologia/Tecnología/Technology'
DIAMETER = 'Diàmetre comptador (cm)/Diámetro contador (cm)/Counter diameter (cm)'
USAGE = 'Ús/Uso/Use'
HOUSING = "Tipus d'habitatge/Tipo de vivienda/Type of housing"
DATE = 'Data/Fecha/Date'
CONSUMPTION = 'Índex de lectura (L/h)/Índice de lectura (L/h)/Reading index (L/h)'
```

```python
import pandas as pd
import pyarrow.dataset as ds
from sklearn.model_selection import GroupShuffleSplit

file_path = '../data/lectures_horaries_ABD.parquet'

dataset = ds.dataset(file_path, format="parquet")
table = dataset.to_table()
df = table.to_pandas()
```

```python
# Initialize GroupShuffleSplit for training and temporary sets
gss = GroupShuffleSplit(n_splits=1, train_size=0.7, random_state=42)
train_idx, temp_idx = next(gss.split(df, groups=df[POLICY]))

train_data = df.iloc[train_idx]
temp_data = df.iloc[temp_idx]

# Split temp_data for validation and testing
gss_val_test = GroupShuffleSplit(n_splits=1, train_size=0.5, random_state=42)
val_idx, test_idx = next(gss_val_test.split(temp_data, groups=temp_data[POLICY]))

validation_data = temp_data.iloc[val_idx]
test_data = temp_data.iloc[test_idx]

# Resulting datasets
print("Training Data:")
print(train_data.head())
print("\nValidation Data:")
print(validation_data.head())
print("\nTest Data:")
print(test_data.head())
```
